##Prueba de Robustez

1. **Independencia del repositorio:** El módulo acepta cualquier ruta local o URL.
2. **Soporte nativo de URLs:** La función `cargar_imagen()` detecta
   automáticamente si recibe una URL (`http://` o `https://`) y la
   descarga usando `requests` + `BytesIO`, sin necesidad de archivos locales.
3. **Soporte de archivos subidos manualmente:** Si el usuario arrastra
   una imagen a Colab, el código la procesa igual que una del repo.
4. **Validaciones:** Si la URL es inválida o el
   archivo no existe, se lanza una `ValueError` controlada en lugar de
   un crash.


```python
print("="*60)
print("Descarga y procesamiento desde URL")
print("="*60)
print("La función cargar_imagen() detecta automáticamente http/https")
print("y la descarga en memoria usando requests + BytesIO.\n")

# URL de ejemplo
url_ejemplo = "https://raw.githubusercontent.com/chrismoffardin/Proyecto_2_Moreno-Moffardin/refs/heads/main/fotos_originales/imagen_clara.jpg"

# Input interactivo: permite al usuario pegar su propia URL o usar la de ejemplo
url_ingresada = input(f"Ingrese una URL de imagen (o presione ENTER para usar la de ejemplo):\n[{url_ejemplo}]\n> ").strip()

# Si el usuario no escribe nada, usa la del ejemplo
if not url_ingresada:
    url_ingresada = url_ejemplo

print("\n" + "="*60)
print(f"📥 Procesando URL: {url_ingresada}")
print("="*60)

try:
    # 1. Intentamos cargar la imagen desde la URL
    fotoapp.cargar_imagen_sistema(url_ingresada)
    print("✅ ¡Imagen descargada y cargada en memoria exitosamente!")

    # 2. Demostramos que el sistema la trata igual que una imagen local
    print("\n👉 Aplicando un proceso (ej: Ajuste de Contraste) para demostrar que el flujo completo funciona...")
    fotoapp.ajustar_contraste()

    print("\n🎉 Prueba completada con éxito. La imagen externa fue procesada igual que una local.")

except ValueError as e:
    # Si la URL es inválida o falla la red, el sistema lo maneja con elegancia
    print("\n⚠️ El sistema detectó un problema y lo manejó de forma segura (no se colgó):")
    print(f"   Mensaje de error controlado: {e}")
    print("   (Esto demuestra que las excepciones están funcionando).")
```

```python
print("="*60)
print("ESCENARIO B: Procesamiento de archivo local subido a Colab")
print("="*60)
print("📌 INSTRUCCIONES:")
print("1. En ícono de 📁 'Carpeta' en la barra lateral izquierda de Colab.")
print("2. Arrastra cualquier imagen .jpg o .png a esta carpeta.")
print("3. Escribe el nombre exacto del archivo abajo (ej: 'mi_foto.jpg').")
print("   (O presiona ENTER para omitir esta prueba si no subiste nada).\n")

nombre_archivo = input("Nombre del archivo subido (con extensión): ").strip()

if nombre_archivo:
    try:
        print(f"\n📂 Leyendo archivo local: {nombre_archivo}")
        fotoapp.cargar_imagen_sistema(nombre_archivo)

        # Demostramos que funciona con OTRA función distinta
        print("👉 Aplicando detección de bordes para demostrar versatilidad")
        fotoapp.detectar_bordes()
        print("✅ ¡Procesamiento de archivo local exitoso!\n")

    except ValueError as e:
        print(f"\n❌ Error controlado: No se pudo procesar el archivo.")
        print(f"   Detalle: {e}")
        print("   (Verifica que el nombre esté bien escrito y que el archivo esté en la carpeta actual).")
else:
    print("⏭️ Prueba de archivo local omitida por el usuario.")

print("\n🎉 Pruebas de robustez finalizadas.")
```
